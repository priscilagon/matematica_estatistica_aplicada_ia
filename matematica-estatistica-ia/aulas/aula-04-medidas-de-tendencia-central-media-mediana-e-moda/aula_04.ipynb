```python
# Passo 0 — baixar o CSV do projeto (executar uma vez)
import os, urllib.request
if not os.path.exists("rio_negro_cotas.csv"):
    urllib.request.urlretrieve("https://raw.githubusercontent.com/priscilagon/matematica-estatistica-ia/main/dados/rio_negro_cotas.csv", "rio_negro_cotas.csv")
print("CSV pronto:", os.path.exists("rio_negro_cotas.csv"))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("rio_negro_cotas.csv", parse_dates=["data"])

# Aula 4 — Média, mediana e moda
x = df["cota_metros"]
print("COTA     média: %.2f | mediana: %.2f" % (x.mean(), x.median()))
print("         (moda de variável contínua quase nunca informa: cada valor aparece ~1 vez)")

c = df["chuva_mm"]
print("CHUVA    média: %.2f | mediana: %.2f | moda: %.1f" %
      (c.mean(), c.median(), c.mode()[0]))

# Moda de variáveis qualitativas (aqui a moda é a medida certa)
print("ALERTA   moda:", df["alerta"].mode()[0])
print("ESPÉCIE  moda:", df["especie_coleta"].mode()[0])

# Efeito de um valor atípico numa janela curta (7 dias em torno de 9/ago)
sem = df[(df["data"] >= "2025-08-06") & (df["data"] <= "2025-08-12")]["cota_metros"]
print("\nSemana de 6 a 12/ago:", list(sem))
print("média: %.2f | mediana: %.2f" % (sem.mean(), sem.median()))

# Medidas por período
print()
print(df.groupby("periodo")["cota_metros"].agg(["mean", "median"]).round(2))
```

## O que observar

- Para a cota (contínua), a moda quase nunca informa: nenhum valor se repete muito.
- Em `chuva_mm`, média ≫ mediana: assimetria à direita.
- `mode()` devolve uma Série (pode haver empate); por isso usamos `.iloc[0]` para pegar a primeira.
- A média da semana de 6 a 12/ago foi “contaminada” pela leitura de 35,10 m; a mediana não.
