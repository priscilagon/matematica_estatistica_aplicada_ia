**Objetos:** Amplitude, variância, desvio-padrão (populacional e amostral).

Compare a variação dos quatro períodos e veja a diferença entre ddof=0 e ddof=1.

> Base **simulada** (didática). Se o download falhar, envie `rio_negro_cotas.csv` pelo painel de arquivos do Colab.

```python
# Passo 0 — baixar o CSV do projeto (executar uma vez)
import os, urllib.request
if not os.path.exists("rio_negro_cotas.csv"):
    urllib.request.urlretrieve("https://raw.githubusercontent.com/priscilagon/matematica-estatistica-ia/main/dados/rio_negro_cotas.csv", "rio_negro_cotas.csv")
print("CSV pronto:", os.path.exists("rio_negro_cotas.csv"))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("rio_negro_cotas.csv", parse_dates=["data"])

# Aula 5 — Amplitude, variância e desvio-padrão (populacional x amostral)
def resumo_disp(s):
    return pd.Series({"n": len(s),
                      "amplitude": s.max() - s.min(),
                      "var_pop": s.var(ddof=0),        # divide por n
                      "var_amostral": s.var(ddof=1),   # divide por n-1
                      "dp_pop": s.std(ddof=0),
                      "dp_amostral": s.std(ddof=1)})

ordem = ["enchente", "cheia", "vazante", "seca"]
tab = (df.groupby("periodo")["cota_metros"].apply(resumo_disp)
         .unstack().reindex(ordem).round(3))
tab["n"] = tab["n"].astype(int)
print(tab)

# CUIDADO: o pandas usa ddof=1 por padrão; o NumPy usa ddof=0.
x = df.loc[df["periodo"] == "cheia", "cota_metros"]
print("\npandas  .std()      :", round(x.std(), 4))
print("numpy   np.std(x)   :", round(np.std(x), 4))
print("numpy   ddof=1      :", round(np.std(x, ddof=1), 4))
```

## O que observar

- As colunas `dp_pop` e `dp_amostral` são próximas porque n é grande; com n = 5 a diferença seria grande.
- Em enchente e vazante, a amplitude está inflada pelas leituras atípicas (8,90 m e 35,10 m) — compare com o DP.
- As três chamadas finais mostram `0,5937` (pandas e `ddof=1`) e `0,5905` (NumPy padrão).
