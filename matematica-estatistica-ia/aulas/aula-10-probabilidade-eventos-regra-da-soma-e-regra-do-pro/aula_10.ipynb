**Objetos:** Experimentos aleatórios, espaço amostral, eventos. Regra da soma e do produto. Evento complementar.

Estime P(A), P(B), P(A e B), a regra da soma e teste a independência com contagens.

```python
# Passo 0 — baixar o CSV do projeto (executar uma vez)
import os, urllib.request
if not os.path.exists("rio_negro_cotas.csv"):
    urllib.request.urlretrieve("https://raw.githubusercontent.com/priscilagon/matematica-estatistica-ia/main/dados/rio_negro_cotas.csv", "rio_negro_cotas.csv")
print("CSV pronto:", os.path.exists("rio_negro_cotas.csv"))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("rio_negro_cotas.csv", parse_dates=["data"])

# Aula 10 — Probabilidade básica, eventos, regra da soma e do produto
n = len(df)
A = df["alerta"] == "alto"          # evento A: cota em nível alto
B = df["chuva_mm"] > 0              # evento B: choveu no dia

p_A, p_B = A.mean(), B.mean()
p_AeB = (A & B).mean()
print("P(A)   = %.4f   (%d de %d dias)" % (p_A, A.sum(), n))
print("P(B)   = %.4f" % p_B)
print("P(A e B) = %.4f" % p_AeB)

# Regra da soma: P(A ou B) = P(A) + P(B) - P(A e B)
print("\nRegra da soma  : %.4f" % (p_A + p_B - p_AeB))
print("Contagem direta: %.4f" % (A | B).mean())

# Regra do produto SE fossem independentes: P(A e B) = P(A) * P(B)
print("\nP(A)*P(B)      : %.4f  <- compare com P(A e B) = %.4f" % (p_A * p_B, p_AeB))
print("Dias com A e B: esperado se independentes = %.1f | observado = %d" %
      (n * p_A * p_B, (A & B).sum()))

# Evento complementar
print("\nP(não A) = %.4f" % (1 - p_A))
```

## O que observar

- A média de uma coluna booleana (`True/False`) é a **proporção** de `True`: é a forma mais simples de estimar probabilidades.
- A regra da soma e a contagem direta dão o mesmo valor (0,7096).
- O produto P(A)×P(B) = 0,1830 é próximo de P(A e B) = 0,1671, mas não igual.
