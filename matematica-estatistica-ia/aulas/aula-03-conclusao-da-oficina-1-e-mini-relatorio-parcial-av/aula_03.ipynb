```python
# Passo 0 — baixar o CSV do projeto (executar uma vez)
import os, urllib.request
if not os.path.exists("rio_negro_cotas.csv"):
    urllib.request.urlretrieve("https://raw.githubusercontent.com/priscilagon/matematica-estatistica-ia/main/dados/rio_negro_cotas.csv", "rio_negro_cotas.csv")
print("CSV pronto:", os.path.exists("rio_negro_cotas.csv"))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("rio_negro_cotas.csv", parse_dates=["data"])

# Aula 3 — Mini-relatório parcial (AV1): tabela-resumo + gráfico para o relatório
ordem = ["enchente", "cheia", "vazante", "seca"]

resumo = (df.groupby("periodo")["cota_metros"]
            .agg(dias="count", minimo="min", maximo="max")
            .reindex(ordem))
print(resumo)

# gráfico de linha com a cota diária e o rótulo de cada período
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(df["data"], df["cota_metros"], color="#2a78d6", linewidth=1.5)
ax.set_title("Cota diária do Rio Negro — Porto de Manaus, 2025 (dados didáticos)")
ax.set_ylabel("Cota (m)")
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
fig.savefig("grafico_cota.png", dpi=150)   # arquivo para colar no relatório
plt.show()
```

## O que observar

- `groupby("periodo")` separa os dados por período; `.agg(...)` calcula várias medidas de uma vez.
- `reindex(ordem)` coloca os períodos na ordem do ano, e não em ordem alfabética.
- `fig.savefig(...)` grava a imagem; use `dpi=150` para boa qualidade de impressão.
