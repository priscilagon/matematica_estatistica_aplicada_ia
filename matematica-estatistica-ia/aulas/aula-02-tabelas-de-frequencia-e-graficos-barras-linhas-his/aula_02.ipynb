**Objetos:** Tabelas e gráficos (barras, linhas, histogramas, boxplot).

A célula abaixo monta a tabela de frequência das espécies, agrupa a cota em classes e desenha quatro gráficos de uma vez.

> Base **simulada** (didática). Se o download falhar, envie `rio_negro_cotas.csv` pelo painel de arquivos do Colab.

```python
# Passo 0 — baixar o CSV do projeto (executar uma vez)
import os, urllib.request
if not os.path.exists("rio_negro_cotas.csv"):
    urllib.request.urlretrieve("https://raw.githubusercontent.com/priscilagon/matematica-estatistica-ia/main/dados/rio_negro_cotas.csv", "rio_negro_cotas.csv")
print("CSV pronto:", os.path.exists("rio_negro_cotas.csv"))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("rio_negro_cotas.csv", parse_dates=["data"])

# Aula 2 — Tabelas de frequência e gráficos
# 1) Variável nominal: espécies registradas
freq = df["especie_coleta"].value_counts()
tabela = pd.DataFrame({"freq_abs": freq,
                       "freq_rel_%": (freq / freq.sum() * 100).round(1)})
print(tabela)
print()

# 2) Variável contínua: agrupar em classes
classes = pd.cut(df["cota_metros"], bins=[8, 12, 16, 20, 24, 28, 36])
print(classes.value_counts().sort_index())

# 3) Quatro gráficos, cada um para um tipo de pergunta
fig, ax = plt.subplots(2, 2, figsize=(11, 7))
freq.plot(kind="bar", ax=ax[0, 0], title="Espécies registradas (barras)")
ax[0, 1].plot(df["data"], df["cota_metros"])
ax[0, 1].set_title("Cota ao longo de 2025 (linha)")
ax[1, 0].hist(df["cota_metros"], bins=12, edgecolor="white")
ax[1, 0].set_title("Distribuição da cota (histograma)")
df.boxplot(column="cota_metros", by="periodo", ax=ax[1, 1])
ax[1, 1].set_title("Cota por período (boxplot)")
plt.suptitle("")
plt.tight_layout()
plt.show()
```

## O que observar

- `value_counts()` já conta as categorias; dividir por `freq.sum()` dá a frequência relativa.
- `pd.cut` cria as classes da cota; o intervalo `(16, 20]` inclui 20 e exclui 16.
- Cada gráfico responde a uma pergunta diferente — veja a tabela acima.
